```python
import pandas as pd

df = pd.read_csv('aice_day5_practice.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['age'] = df['age'].fillna(df['age'].mean())
df['monthly_fee'] = df['monthly_fee'].fillna(df['monthly_fee'].median())
df['usage_gb'] = df['usage_gb'].fillna(df['usage_gb'].mean())
df['city'] = df['city'].fillna(df['city'].mode().iloc[0])
df['plan'] = df['plan'].fillna(df['plan'].mode().iloc[0])
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df['gender'] = df['gender'].map({'M':0,'F':1})
```

```python
df['plan'].value_counts()
```

```python
df = pd.get_dummies(df, columns=['plan'])
```

```python
df['contract_type'].value_counts()
```

```python
df = pd.get_dummies(df, columns=['contract_type'])
```

```python
df['city'].value_counts()
```

```python
df = pd.get_dummies(df, columns=['city'])
```

```python
df.info()
```

```python
df.head()
```

```python
x = df.drop(['churn', 'customer_id'], axis=1)
y =df['churn']
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test= train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

model = DecisionTreeClassifier(random_state=42)

model.fit(x_train, y_train)
train_pred = model.predict(x_train)
test_pred = model.predict(x_test)

train_acc = accuracy_score(y_train, train_pred)
test_acc = accuracy_score(y_test, test_pred)

print(train_acc, test_acc) 
# 1.0 / 0.8333... 
# max_depth를 주지 않은 상태에서 계속 분기하면서 과적합이 됐을 가능성이 높다
```

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(
    model,
    x_train,
    y_train,
    cv=5,
    scoring='accuracy'
)

print(scores)
print(scores.mean())
```

```python
from sklearn.model_selection import GridSearchCV

# 최적의 max_depth 하이퍼파라미터 찾기
params = {
    'max_depth': [2, 3, 4, 5, 6, 7, 8, 10]
}

grid = GridSearchCV(
    model,
    params,
    cv=5, # 총 40회의 cv학습이 이루어짐. 기본이 refit=true 이기 때문에 +1 -> 41회
    scoring='accuracy'
)

grid.fit(x_train, y_train)

print(grid.best_params_)
print(grid.best_score_)
# 이미 최적 파라미터로 다시 학습된 모델
print(grid.best_estimator_)
```

```python
# 최적 파라미터로 다시 학습된 모델로 테스트
best_model = grid.best_estimator_

final_pred = best_model.predict(x_test)

final_acc = accuracy_score(y_test, final_pred)

final_acc
```

```python
params2 = {
    'max_depth':[2, 3, 4, 5],
    'min_samples_split':[2, 5, 10],
    'min_samples_leaf':[1, 2, 4]
}

grid_multi = GridSearchCV(
    model,
    params2,
    cv=5,
    scoring='accuracy'
)

grid_multi.fit(x_train, y_train)

print(grid_multi.best_params_)
print(grid_multi.best_score_)
print(grid_multi.best_estimator_)
```

```python
best_model2 = grid_multi.best_estimator_

train_pred2 = best_model2.predict(x_train)
test_pred2= best_model2.predict(x_test)

train_acc2 = accuracy_score(y_train, train_pred2)
test_acc2 = accuracy_score(y_test, test_pred2)

print(train_acc2)
print(test_acc2)
```

```python
# 분류 모델 성적표
from sklearn.metrics import classification_report

print(classification_report(y_test, test_pred2))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, test_pred2)
cm
```

```python
grid_multi2 = GridSearchCV(
    model,
    params2,
    cv=5,
    scoring='recall' # recall 기준
)

grid_multi2.fit(x_train, y_train)

test_pred3 = grid_multi2.predict(x_test)

report = classification_report(y_test, test_pred3)
cm2 = confusion_matrix(y_test, test_pred3)

print(report)
print(cm2)
```

```python
print(grid_multi2.best_params_) # recall 기준 시 다른 모델을 고름
print(grid_multi2.best_score_)
```

```python
# RandomizedSearchCV - 랜덤한 조합으로 검사
from sklearn.model_selection import RandomizedSearchCV

params_random = {
    'max_depth': [2, 3, 4, 5, 6, 7, 8, 9, 10],
    'min_samples_split': [2, 3, 4, 5, 6, 8, 10, 15],
    'min_samples_leaf': [1, 2, 3, 4, 5, 6, 8]
}

random_search = RandomizedSearchCV(
    model,
    params_random,
    n_iter=20,
    cv=5,
    scoring='accuracy',
    random_state=42
)

random_search.fit(x_train, y_train);
```

```python
print(random_search.best_score_)
print(random_search.best_params_)
print(random_search.best_estimator_)
```
