```python
flight = pd.read_csv("C:/Users/LSG/Desktop/AICE_자료/Clean_Dataset.csv/Clean_Dataset.csv", encoding='cp949')
flight.head()
```

```python
# 컬럼명 변경하기 - 딕셔너리처럼 변경
flight = flight.rename(columns={'airline':'airline_name', 'source_city':'departure_city'})
flight.head()
```

```python
# 내림차순 정렬
flight=flight.sort_values(by='Unnamed: 0', ascending=False)
# 오름차순 정렬
flight=flight.sort_values(by='Unnamed: 0', ascending=True)
flight
```

```python
flight = flight.rename(columns={'airline_name':'airline'})
# groupby 활용
airline_group = flight.groupby('airline')
airline_group
```

```python
# 해당 방식으로 그룹화된 데이터를 조회 가능
airline_group.groups
```

```python
# airline 별 데이터 수 확인
airline_group.count()
```

```python
# 숫자 컬럼만 골라서 평균
airline_group.mean(numeric_only=True)
```

```python
# 특정 컬럼 값만 확인하기
airline_group.mean(numeric_only=True)[['price']]
```

```python
# 다중 인덱싱
# .mean(numeric_only=True) 를 따로 붙여도 가능
airline_group2 = flight.groupby(['airline', 'arrival_time']).mean(numeric_only=True)
airline_group2
```

```python
airline_group2.loc[[('AirAsia','Evening')]]
```

```python
# 기존 컬럼 인덱스로 설정하기
flight_index = flight.set_index(['airline','arrival_time'])
flight_index
```

```python
# 다중 인덱스 세팅 후 해당 인덱스 기준으로 groupby하기 
# level[0,1] =  인덱스 0번째 레벨과 1번째 레벨을 기준으로 그룹화 - 다중 인덱스의 순서 번호 
flight_index2 = flight.set_index(['airline','arrival_time']).groupby(level=[0,1]).mean(numeric_only=True)
flight_index2
```

```python
# Aggregate로 집계하기 
# 데이터 프레임의 값을 다양하게 집계하여 한 번에 볼 수 있음
agg_index = flight.set_index(['airline', 'arrival_time']).groupby(level=[0,1]).aggregate([np.min, np.max]).head()
agg_index
```
