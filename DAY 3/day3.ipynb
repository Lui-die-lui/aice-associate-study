```python
for depth in range(1,11):
    range_model = DecisionTreeClassifier(
        max_depth=depth,
        random_state=42
    )

    range_model.fit(x_train, y_train)
    pred_depth = range_model.predict(x_test)

    acc = accuracy_score(y_test, pred_depth)

    print(depth, acc)
```

```python
from sklearn.model_selection import cross_val_score

cv_score = cross_val_score(
    DecisionTreeClassifier(max_depth=4, random_state=42),
    x_train,
    y_train,
    cv=5,
    scoring='accuracy'
)

cv_score
```

```python
cv_score.mean()
```

```python
best_depth = None
best_score = 0

for depth in range(1, 11):

    # 모델 생성
    range_model = DecisionTreeClassifier(
        max_depth=depth,
        random_state=42
    )

    # cross validation
    scores = cross_val_score(
        range_model,
        x_train,
        y_train,
        cv=5,
        scoring='accuracy'
    )

    mean_score = scores.mean()

    # best_score와 현재 평균 점수를 비교해서 best_score에 큰 값을 넣는다. 해당 depth 도 가장 최적으로 분류되어 대입시켜준다.
    if mean_score > best_score:
        best_score = mean_score
        best_depth = depth

print('best_depth :', best_depth)
print('best_score :', best_score)

```

```python
# final_tree 학습
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

final_model = DecisionTreeClassifier(
    max_depth=6,
    random_state=42
)

final_model.fit(x_train, y_train)
final_pred = final_model.predict(x_test)

final_accuracy = accuracy_score(y_test, final_pred)

final_accuracy

```
