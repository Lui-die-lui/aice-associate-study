```python
print(df['city'].value_counts(), '\n')
print(df['weather'].value_counts(dropna=False), '\n')
print(df['traffic_level'].value_counts(), '\n')
print(df['vehicle_type'].value_counts(), '\n')
print(df['order_channel'].value_counts(), '\n')
```

```python
df['weather'] = df['weather'].fillna(df['weather'].mode().iloc[0])
df['distance_km'] = df['distance_km'].fillna(df['distance_km'].median())
df['courier_experience_years'] = df['courier_experience_years'].fillna(df['courier_experience_years'].mean())
```

```python
df.isnull().sum()
```

```python
# 날짜 처리 
df['order_date'] = pd.to_datetime(df['order_date'])
df['order_date'].dtype
```

```python
# 년/월/일/요일/주말 날짜 컬럼
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['day'] = df['order_date'].dt.day
df['dayofweek'] = df['order_date'].dt.day_of_week
df['is_weekend'] = (df['dayofweek']>=5).astype(int)

df[['order_date', 'year', 'month', 'day', 'dayofweek', 'is_weekend']].head()
```

```python
df['traffic_level'] = df['traffic_level'].map({
    'Medium':0,
    'Low':1,
    'High':2
})
```

```python
# 이미 들어있는 컬럼 삭제
df.drop(columns=['order_date', 'year', 'day'], inplace=True)
```

```python
df = pd.get_dummies(df, columns=['city', 'weather', 'vehicle_type', 'order_channel'])
```

```python
df = df.drop(columns=['order_id'])
```

```python
# x / y 분리
from sklearn.model_selection import train_test_split

x = df.drop(columns=['delivery_minutes'])
y = df['delivery_minutes']

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
# 스케일링
# 평균 기준으로 얼마나 떨어져 있는지 보여줘서 음/양수 섞임
from sklearn.preprocessing import StandardScaler

st_scale = StandardScaler()

st_x_train_scaled = st_scale.fit_transform(x_train)
st_x_test_scaled = st_scale.transform(x_test)
```

```python
# 전부 0~1 범위 쪽으로 누름
from sklearn.preprocessing import MinMaxScaler

mm_scale = MinMaxScaler()

mm_x_train_scaled = mm_scale.fit_transform(x_train)
mm_x_test_scaled = mm_scale.transform(x_test)
```

```python
# 중앙값과 IQR 기준이라 값 분포가 다르게 보임
from sklearn.preprocessing import RobustScaler

rb_scale = RobustScaler()

rb_x_train_scaled = rb_scale.fit_transform(x_train)
rb_x_train_scaled = rb_scale.transform(x_test)
```

```python
# 스케일러 별 비교 (5개행씩)
print(st_x_train_scaled[:5, 1])
print(mm_x_train_scaled[:5, 1])
print(rb_x_train_scaled[:5, 1])
x_train.columns.get_loc('distance_km')
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(x_train, y_train)

pred = lr.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, pred)
mae
```

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, pred)
print(mse)
# rmse - mse를 분으로 변환
print(np.sqrt(mse))
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, pred)

r2
```
